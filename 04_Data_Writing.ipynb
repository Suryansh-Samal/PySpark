```python
df = spark.read.format("csv").option("inferSchema", True).option("header", True).load("/Volumes/workspace/default/pyspark_data/BigMart Sales.csv")
```

```python
df.limit(10).display()
```

### CSV

```python
df.write.format("csv")\
    .save("/Volumes/workspace/default/pyspark_data/output/csv")
```

### Append

```python
df.write.format("csv")\
    .mode("append")\
        .option("header", True)\
             .option("path", "/Volumes/workspace/default/pyspark_data/output/csv")\
                 .save()
```

### Overwrite

```python
df.write.format("csv")\
    .mode("overwrite")\
        .option("header", True)\
            .option("path", "/Volumes/workspace/default/pyspark_data/output/csv" )\
                .save()
       
```

### Error

```python
df.write.format("csv")\
    .mode("error")\
        .option("header", True)\
            .option("path", "/Volumes/workspace/default/pyspark_data/output/csv")\
                .save()
```

### Ignore

```python
df.write.format("csv")\
    .mode("ignore")\
        .option("header", True)\
            .option("path", "/Volumes/workspace/default/pyspark_data/output/csv")\
                .save()
```

### Parquet

```python
df.write.format("parquet")\
    .mode("overwrite")\
        .option("path", "/Volumes/workspace/default/pyspark_data/output/csv")\
             .save()
```

### Table

```python
df.write.format("delta")\
    .mode("overwrite")\
        .saveAsTable("my_table")
```
