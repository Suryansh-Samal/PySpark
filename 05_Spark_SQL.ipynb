```python
df = spark.read.format("csv")\
    .option("inferSchema", True)\
        .option("header", True)\
            .load("/Volumes/workspace/default/pyspark_data/BigMart Sales.csv")
                                 
```

### create TempView

```python
df.createTempView("my_view")
```

```python
%sql
SELECT *
FROM my_view
LIMIT 10;
```

```python
%sql
Select * from my_view
where Item_fat_content = 'Regular'
limit 10;
```

### Writing Using SQl

```python
df_sql = spark.sql("Select * from my_view where Item_Fat_content = 'Regular' limit 10 ")
```

```python
df_sql.display()
```

```python
df_sql.write.format("csv")\
    .mode("append")\
        .option("header", True)\
            .option("path", "/Volumes/workspace/default/pyspark_data/output/csv")\
                .save()
```
